```python
# ---- 1. Rename ----
```

```python
df = df.rename(columns={"Name":"employee_name"})
print(df.columns.tolist())
```

```python
# ---- 2. Fix data types ----
```

```python
df["hire_date"] = pd.to_datetime(df["hire_date"],errors="coerce")
df["hire_date"].isna().sum()
```

```python
df['DOB'].dtype #check type 
```

```python
df["DOB"] = pd.to_datetime(df["DOB"],errors="coerce")
df["exit_date"] = pd.to_datetime(df["exit_date"], errors="coerce",dayfirst=True)
print(df["DOB"].isna().sum())
```

```python
# ---- 3. Standardize text/categorical casing FIRST (before filling, so fills match clean categories) ----
df["employee_name"] = df["employee_name"].str.strip().str.title()
df["attrition"] = df["attrition"].str.strip().str.title()
df["gender"] = df["gender"].replace({"M": "Male", "m": "Male", "MALE": "Male",
                                       "F": "Female", "f": "Female", "FEMALE": "Female"})
df["overtime"] = df["overtime"].replace({"Y": "Yes", "yes": "Yes", "N": "No", "no": "No"})
df["marital_status"] = df["marital_status"].str.strip().str.title()
df["business_travel"] = df["business_travel"].str.strip().str.title()
```

```python
# ---- 4. Fix invalid numeric values (business-logic errors) ----
```

```python
median_age = df.loc[(df["age"] > 18) & (df["age"] < 65), "age"].median()
df["age"] = df["age"].apply(lambda x: median_age if (x <= 18 or x >= 65) else x)
```

```python
df["performance_rating"] = df["performance_rating"].where(df["performance_rating"].between(1,5))
df["monthly_salary"] = df["monthly_salary"].where(df["monthly_salary"] > 0)
```

```python
# ---- 5. Fill missing values — grouped/logical imputation ----
```

```python
df["monthly_salary"] = df["monthly_salary"].fillna(
    df.groupby("job_role")["monthly_salary"].transform("median"))
```

```python
df["education_field"] = df["education_field"].fillna(
    df.groupby("job_role")["education_field"].transform(lambda x: x.mode()[0] if not x.mode().empty else "Unknown"))
```

```python
df["distance_from_home"] = df["distance_from_home"].fillna(
    df.groupby("city")["distance_from_home"].transform("median"))
```

```python
df["performance_rating"] = df["performance_rating"].fillna(df["performance_rating"].median())
```

```python
df["training_times_last_year"] = df["training_times_last_year"].fillna(df["training_times_last_year"].median())
```

```python
df["num_companies_worked"] = df["num_companies_worked"].fillna(df["num_companies_worked"].median())
```

```python
df["overtime"] = df["overtime"].fillna(df["overtime"].mode()[0])
df["business_travel"] = df["business_travel"].fillna(df["business_travel"].mode()[0])
```

```python
# ---- 6. Fill values that genuinely can't be recovered — mark explicitly, don't guess ----
```

```python
df["gender"] = df["gender"].fillna("Not Disclosed")
```

```python
df["marital_status"] = df["marital_status"].fillna("Not Disclosed")
```

```python
# ---- 7. Leave subjective survey scores as-is, but flag rows missing them ----
```

```python
survey_cols = ["job_satisfaction", "environment_satisfaction", "work_life_balance"]
df["missing_survey_data"] = df[survey_cols].isna().any(axis=1)
```

```python
# ---- 9. Remove duplicates ----
```

```python
df = df.drop_duplicates()
df["duplicate_emp_id"] = df["employee_id"].duplicated(keep = False)
```

```python
# ---- 10. Feature engineering ----
```

```python
df["age_group"] = pd.cut(df["age"], bins=[18, 25, 35, 45, 65], labels=["18-25", "26-35", "36-45", "46-65"])
df["tenure_group"] = pd.cut(df["years_at_company"], bins=[-1, 1, 3, 6, 20], labels=["<1yr", "1-3yr", "3-6yr", "6+yr"])
```

```python
# ---- 11. Final check ----
print("Remaining missing values:\n", df.isna().sum())
```

```python
df.isna().sum()
```

```python
df = df.replace(["", " ", "null", "NULL", "None", "N/A", "NaN", "nan"], np.nan)
```

```python
df["gender"] = df["gender"].fillna("Not Disclosed")
df["marital_status"] = df["marital_status"].fillna("Not Disclosed")
```

```python
df["age"] = df["age"].apply(lambda x: median_age if pd.isna(x) or x <= 18 or x >= 65 else x)
```
